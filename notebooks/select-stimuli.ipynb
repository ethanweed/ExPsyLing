```python
# import libraries
import pandas as pd
import random

# define paths and filenames
pathin = "/Users/ethan/Documents/GitHub/ExPsyLing/labjs exercises/LDT replication/"
pathout = "/Users/ethan/Documents/GitHub/ExPsyLing/labjs exercises/LDT replication/"

infile = "all ldt subs_all trials.csv"
outfile = "stimuli.csv"

# read in the raw data
raw = pd.read_csv(pathin + infile, low_memory=False)  #the low_memory part is just to suppress an irrelevant warning

# Find stimuili pairs for the three conditions: related, unrelated, and nonword
df_rel = raw[(raw['rel'] == 'rel') & (raw['type'] == 'first')][['prime', 'target']]
df_un = raw[(raw['rel'] == 'un') & (raw['type'] == 'first')][['prime', 'target']]
df_nw = raw[raw['type'] == 'nw'][['prime', 'target']]

# Create a new column that combines the prime and target into a single string, so we can easily identify unique pairs
df_rel['combined'] = df_rel['prime'] + " " + df_rel['target']
df_un['combined'] = df_un['prime'] + " " + df_un['target']
df_nw['combined'] = df_nw['prime'] + " " + df_nw['target']

# Create sets of unique pairs for each condition
rel = set(list(df_rel['combined']))
un = set(list(df_un['combined']))
nw = set(list(df_nw['combined']))

# Set a seed (place for the random number generator to start). This way you will always get the same "random" order.
# If you don't have a seed, you will get a different random order each time you run the code.
# Lots of programmers use 42, because they are nerds, but you can pick a different number if you like.
random.seed(42)   

# Convert the sets to lists so we can shuffle them
rel = list(rel)
un = list(un)
nw = list(nw)

# Shuffle the lists so that we can randomly select stimuli from them
random.shuffle(rel)
random.shuffle(un)
random.shuffle(nw)

# Select the first 20 stimuli from each condition
rel_stimuli = rel[:20]
un_stimuli = un[:20]
nw_stimuli = nw[:20]

# Now that we have picked stimulus pairs, re-split them so they can be in each their column
rel_targets = [stim.split(" ")[1] for stim in rel_stimuli]
rel_primes = [stim.split(" ")[0] for stim in rel_stimuli]
un_targets = [stim.split(" ")[1] for stim in un_stimuli]
un_primes = [stim.split(" ")[0] for stim in un_stimuli]
nw_targets = [stim.split(" ")[1] for stim in nw_stimuli]
nw_primes = [stim.split(" ")[0] for stim in nw_stimuli]

# make a list with all the primes, and a list with all the targets
primes = rel_primes + un_primes + nw_primes
targets = rel_targets + un_targets + nw_targets

# make a list with the condition labels for each stimulus pair
conditions = ['related'] * len(rel_stimuli) + ['unrelated'] * len(un_stimuli) + ['nonword'] * len(nw_stimuli)

# Create a new dataframe with the selected stimuli and their conditions
df_stimuli = pd.DataFrame({'prime': primes, 'target': targets, 'condition': conditions})

# Save the new dataframe to a CSV file, for importing into labjs
df_stimuli.to_csv(pathout + outfile, index=False)

```
