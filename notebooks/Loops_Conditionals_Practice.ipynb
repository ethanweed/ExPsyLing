## `while` loops

[LSP 8.2.1 While loops](https://ethanweed.github.io/pythonbook/03.04-basic_programming.html#while-loops): a `while` loop keeps running *as long as* some condition stays true. It's a good fit when you don't know in advance how many times you'll repeat -- you just know when to stop.

**Live-code demo** (from the book):

```python
sweet_enough = 3
num_sugar_spoons = 0
while num_sugar_spoons < sweet_enough:
    num_sugar_spoons = num_sugar_spoons + 1
    print('I have added', num_sugar_spoons, 'spoons of sugar')
```

## `for` loops

[LSP 8.2.2 For loops](https://ethanweed.github.io/pythonbook/03.04-basic_programming.html#for-loops): a `for` loop runs once for every item in a collection -- a list, a string, a `dict`'s keys, whatever you hand it. Where `while` asks "keep going until X", `for` says "do this once per item, then stop."

A quick note on syntax ([LSP 8.2.3](https://ethanweed.github.io/pythonbook/03.04-basic_programming.html#loop-syntax)): every loop header ends in a colon, and everything that belongs *inside* the loop has to be indented (4 spaces, consistently). Most IDEs handle the indentation for us, but keep an eye on it. If you get an error, that might be why. The loop variable name (`egg`, `word`, whatever you pick) is entirely up to you -- Python doesn't care what you call it, only that you use the same name inside the loop body.

**Live-code demo** (from the book):

```python
eggs = ['egg', 'egg', 'egg']
for egg in eggs:
    print('I cracked an egg!')
```

```python
words = ['eggs', 'flour', 'sugar', 'cinnamon', 'salt']
for word in words:
    print('There are', len(word), 'letters in', word.upper())
```

## Conditionals: `if` / `elif` / `else`

[LSP 8.3 Conditional statements](https://ethanweed.github.io/pythonbook/03.04-basic_programming.html#conditional-statements): `if` runs a block only when a condition is true; `elif` ("else if") checks another condition if the first one wasn't true; `else` catches everything left over. You can chain as many `elif`s as you like, but `else` (if you use one) always comes last, with no condition of its own.

**Live-code demo** (from the book -- note this one combines a `for` loop with `if`/`elif`/`else`), but you don't always have to combine all of these:

```python
week = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
for day in week:
    if day == 'Wednesday':
        print('Today is', day + '.', 'You need to leave early!')
    elif day == 'Saturday' or day == 'Sunday':
        print('Today is', day + '.', 'You can sleep late!')
    else:
        print('Today is', day + '.', 'You can leave at the normal time today.')
```

## Functions

[LSP 8.4 Functions](https://ethanweed.github.io/pythonbook/03.04-basic_programming.html#functions): a function packages up a block of code so you can run it again, on different inputs, without retyping it. `def` names the function and lists its parameters; everything indented underneath is the function body. To run the function, you need to call it by name (with arguments, if necessary).

**Live-code demo** (from the book):

```python
def measure_word(word):
    letter_count = 0
    for letter in word:
        letter_count = letter_count + 1
    print('There are', letter_count, 'letters in', word.upper())

measure_word('jackalope')
```

Can you write a shorter function that does the same thing, but uses `len()` instead?

```python

def word_measurer(word):
    num_letters = len(word)
    return(num_letters)
```

```python
x = word_measurer('Friday')
```

```python
x
```

**Saving function output**: Often (more like, usually) we want to get a value out of a function, but we don't just want to print the answer; we want to save it for later, to use in some other context. For this, we can use `return` instead of `print` in our function.

```python
def measure_word(word):
    letter_count = 0
    for letter in word:
        letter_count = letter_count + 1
    return(letter_count)

output = measure_word('jackalope')
```

```python
print(output)
```

```python
jackalope = measure_word('jackalope')
snipe = measure_word('snipe')
total_letters = jackalope + snipe
print(total_letters)
```

## List comprehensions: a different kind of loop

[LSP 8.5 List comprehensions](https://ethanweed.github.io/pythonbook/03.04-basic_programming.html#list-comprehensions-a-different-kind-of-loop): a list comprehension is a `for` + `if` + build-a-new-list loop, squeezed onto one line: `[expression for item in collection if condition]`. List comprehensions can feel very confusing, but in principle they are the exact same loop you already know how to write, just compressed. The `if` part is optional if you're not filtering anything.

**Live-code demo** (from the book):

```python
words = ['If', 'you', 'say', 'it', 'loud', 'enough', 'you', 'will', 'always', 'sound', 'precocious']
new_words = [x for x in words if 'o' in x]
print(new_words)
```

```python
long_words = ['_' + x + '_' for x in words if len(x) > 4]
print(long_words)
```

Each level of nesting gets its own extra level of indentation.

### Nested conditionals

[LSP 8.6.1](https://ethanweed.github.io/pythonbook/03.04-basic_programming.html#nested-conditionals). **Live-code demo** (from the book):

```python
numbers = [4, 2, 54, 823452, 324, 2, 4435, 4, 9070878072634, 3421, 4345]
for num in numbers:
    if len(str(num)) > 1:
        if num % 2 == 0:
            print(num)
```

## Abstraction, generalization, and patterns

[LSP 8.7](https://ethanweed.github.io/pythonbook/03.04-basic_programming.html#abstraction-generalization-and-patterns): once you notice you're writing the same shape of loop over and over -- a counter, or a filter-and-append -- it's worth turning the specific version into a general-purpose function with a parameter, so you're not stuck rewriting it for every new hardcoded value.

**Live-code demo** (from the book -- a specific filter loop, generalized into a reusable function):

```python
# search for words that begin with 'b'
all_flowers = ['Begonias', 'Lilacs', 'Roses', 'Pansies', 'Foxgloves', 'Buttercups', 'Sunflowers']
b_flowers = []
for flower in all_flowers:
    if flower.startswith('B'):
        b_flowers.append(flower)
print(b_flowers)
```

```python
# search for words that begin with any letter we provide as an argument
def check_first_letter(input_list, target_letter):
    output_list = []
    for item in input_list:
        if item.lower().startswith(target_letter.lower()):
            output_list.append(item)
    print(output_list)

check_first_letter(all_flowers, 'b')
check_first_letter(all_flowers, 's')
```
