# Variables and variable types


An **object** is a thing in memory that holds data and behavior.  
Objects have:  
- Type (e.g., int, str, list)  
- Value (e.g., 42, "hello")  
- Identity (a unique memory address)  

A **variable** is a name that refers to (or points to) an object. 
Its like a label

```python
#This is a comment, meant for people to write down notes, does not affect the output or code

greeting = "hello"
# "Hello" is a string type object
# greeting is a variable

salute = "hello"
# the object does not change but the variable is different

print(greeting)
print(salute)
```

```python
#Python remebers the last version
greeting = "hello"
greeting = "Hi"

print(greeting)
```
