## Indexing & slicing

I didn't do a great job explaining indexing and slicing last week, so let me try again. Also, remember you can consult [LSP 4.17 Indexing and slicing](https://ethanweed.github.io/pythonbook/02.02-more_python_concepts.html#indexing-and-slicing) and [4.18 Slices](https://ethanweed.github.io/pythonbook/02.02-more_python_concepts.html#slices).

In Python, every list (and string, and tuple) is [**0-indexed**](https://en.wikipedia.org/wiki/Zero-based_numbering): the first item is at position `0`, not `1`. Negative indices count backwards from the end, so `-1` is the last item, `-2` is second-to-last, and so on. If it helps, you can think about it in terms of steps. Imagine that the Python list is like a sidewalk, made out of square cement blocks. When you step onto the sidewalk, you are standing on a block. This is your starting position. To get to the block you are already standing on, you would take 0 steps, because you are already there. That first block has an index of 0, which is the number of steps it takes to get there from the start. You take [one step](https://www.youtube.com/watch?v=Lixm0ajWCi8&list=RDLixm0ajWCi8&start_radio=1), and now you are on the second block, so that block gets the index of "1". Maybe this helps?

Now, imagine you are at the end of the sidewalk. You have stepped off of the sidewalk, and you want to get back on. You need to take one step backwards (-1) to get back onto that last block. If you are walking backwards from a position just off of the sidewalk, that last block is at position -1. Does this help? I hope so. The best thing is still to just play around with it and get used to it.

A single index, like `mylist[3]`, gives you back **one item** — whatever type that item is.

A **slice**, written `mylist[start:stop]`, gives you back a **new list** containing everything from position `start` up to *but not including* position `stop`. This last bit, "not including stop", can be quite confusing at first — `mylist[0:3]` gives you 3 items (indices 0, 1, 2), not 4. If it helps, think about Python counting through all the different positions. When it gets to the one in the "stop" position, it say "No, no, no! Stop! Not this one!" In Computational Linguistics you will probably learn about "stop words". These are all the words you _don't_ want to do something with. For example, when analyzing a text, maybe you don't want your analysis to include all of the most common words like "and", "a", "the", so you give your Python script a list of these "stop words", which it should ignore. The stop position in the slice is kind of the same idea.

You can leave either side blank: `mylist[:3]` means "from the beginning up to 3", and `mylist[3:]` means "from 3 to the end".

There's an optional third part, the **step**: `mylist[start:stop:step]` skips items. `mylist[::2]` takes every 2nd item across the whole list. `mylist[::-1]` steps backwards through the whole list.

| Slice | Meaning |
|---|---|
| `mylist[0]` | just the first item (not a list) |
| `mylist[0:1]` | a *list* containing just the first item |
| `mylist[:3]` | first 3 items |
| `mylist[-3:]` | last 3 items |
| `mylist[2:5]` | items at index 2, 3, 4 |
| `mylist[::2]` | every 2nd item |
| `mylist[::-1]` | the whole list, reversed |

### 9. Comparing favorites with a partner

Sets have their own comparison operators, borrowed straight from math-class Venn diagrams:

| Operator | Name | Meaning |
|---|---|---|
| `a & b` | intersection | items in **both** `a` and `b` |
| `a \| b` | union | items in **either** `a` or `b` (combined, no duplicates) |
| `a - b` | difference | items in `a` but **not** in `b` |
| `a ^ b` | symmetric difference | items in **exactly one** of `a` or `b`, not both |
| `x in a` | membership | `True` if `x` is somewhere in `a` |

### 10. Building a dictionary with `zip`

`zip()` takes two (or more) lists and pairs them up position by position: item 1 from the first list with item 1 from the second, item 2 with item 2, and so on. On its own it gives you back a `zip` object, which doesn't print anything useful until you turn it into something else, like a `list()` or a `dict()`:

```python
letters = ["a", "b", "c"]
numbers = [1, 2, 3]

list(zip(letters, numbers))
# [('a', 1), ('b', 2), ('c', 3)]
```

Feed that same pairing straight into `dict()` and each pair becomes a key-value pair instead of a tuple:

```python
dict(zip(letters, numbers))
# {'a': 1, 'b': 2, 'c': 3}
```
